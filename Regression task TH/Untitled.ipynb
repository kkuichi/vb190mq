```python
import pandas as pd

groups = [
    "NCD", "BL", "NAD", "AI", "NAC", "ND",
    "CS", "AT", "NA", "ADL", "AS_NA", "AS_NAC"
]

columns = [f"{group}_{t}" for group in groups for t in range(8)]
columns.append("target")

df = pd.read_csv(
    "TomsHardware.data",
    header=None,
    names=columns
)

print(df.shape)
df.head(20)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df["target"], bins=50)
plt.title("Rozdelenie cieľovej premennej")
plt.xlabel("Cieľová premenná")
plt.ylabel("Frekvencia")
plt.show()
```

```python
df["target"].skew()
```

```python
import numpy as np

plt.figure(figsize=(8,5))
plt.hist(np.log1p(df["target"]), bins=50)
plt.title("Rozdelenie cieľovej premennej po logaritmickej transformácii")
plt.xlabel("log(1 + cieľová premenná)")
plt.ylabel("Frekvencia")
plt.show()
```

```python
df["target_log"] = np.log1p(df["target"])
```
