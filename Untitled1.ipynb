```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
TW = pd.read_csv(
    r"classification/Twitter/Absolute_labeling/Twitter-Absolute-Sigma-500.data",
    sep=",",
    header=None
)

groups = ["NCD", 'AI', 'AS(NA)', 'BL',
         'NAC', 'AS(NAC)', 'CS', 'AT', 'NA','ADL', 'NAD']

columns = []
for group in groups:
    for t in range(7):
        columns.append(f"{group}_{t}")

columns.append("label")  # остання колонка

TW.columns = columns

TW.head(2)
```

```python
df_1 = TW[TW["label"] == 1]
```

```python
cols_ncd = [col for col in TW.columns if "NCD" in col]

df_1[cols_ncd].describe().T
```

```python
plt.figure(figsize=(12,5))
sns.boxplot(data=df_1[cols_ncd])
plt.xticks(rotation=45)
plt.title("NCD features - class 1")
plt.show()
```

```python
cols_ai = [col for col in TW.columns if "AI" in col]

df_1[cols_ai].describe().T
```

```python
plt.figure(figsize=(12,5))
sns.boxplot(data=df_1[cols_ai])
plt.xticks(rotation=45)
plt.title("AI features - class 1")
plt.show()
```
