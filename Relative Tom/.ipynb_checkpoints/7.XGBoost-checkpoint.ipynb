```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from xgboost import XGBClassifier
from sklearn.model_selection import GridSearchCV
from xgboost import XGBClassifier
from sklearn.model_selection import RandomizedSearchCV
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import (accuracy_score,precision_score,recall_score,f1_score,roc_auc_score,confusion_matrix)
```

```python
TH_500 = pd.read_csv(
    "../classification/TomsHardware/Relative_labeling/sigma=500/TomsHardware-Relative-Sigma-500.data",
    sep=",",
    header=None
)

TH_1000= pd.read_csv(
    "../classification/TomsHardware/Relative_labeling/sigma=1000/TomsHardware-Relative-Sigma-1000.data",
    sep=",",
    header=None
)
TH_1500= pd.read_csv(
    "../classification/TomsHardware/Relative_labeling/sigma=1500/TomsHardware-Relative-Sigma-1500.data",
    sep=",",
    header=None
)
groups = [
    "NCD", "BL", "NAD", "AI", "NAC", "ND",
    "CS", "AT", "NA", "ADL", "AS_NA", "AS_NAC"
]

columns = []
for group in groups:
    for t in range(8):
        columns.append(f"{group}_{t}")

columns.append("label")  

TH_500.columns = columns
TH_1000.columns = columns
TH_1500.columns = columns

prefixes = {col.split("_")[0] for col in TH_500.columns if "_" in col}
prefixes = {col.split("_")[0] for col in TH_1000.columns if "_" in col}
prefixes = {col.split("_")[0] for col in TH_1500.columns if "_" in col}
```

### 500

###  1. Baseline XGBoost

```python
X = TH_500.drop(columns=['label'])
y = TH_500['label']

X_train, X_test, y_train, y_test = train_test_split(X, y,test_size=0.2, stratify=y,random_state=42)
```

```python
xgb = XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss').fit(X_train, y_train)


y_pred = xgb.predict(X_test)
y_prob = xgb.predict_proba(X_test)[:,1]

print("1. Baseline XGBoost")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1-score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))

cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
```

### 2. XGBoost with Grid Search

```python
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [3, 6, 10],
    'learning_rate': [0.01, 0.1],
    'subsample': [0.8, 1.0],
    'colsample_bytree': [0.8, 1.0]
}

xgb = XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss')

grid_search_xgb = GridSearchCV( xgb,param_grid,cv=5,scoring='f1',n_jobs=-1).fit(X_train, y_train)

best_xgb = grid_search_xgb.best_estimator_

y_pred = best_xgb.predict(X_test)
y_prob = best_xgb.predict_proba(X_test)[:,1]

print("2. XGBoost with Grid Search")
print("Best parameters:", grid_search_xgb.best_params_)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1-score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```

### 3. XGBoost with scale_pos_weight and Grid Search

```python
neg = (y_train == 0).sum()
pos = (y_train == 1).sum()

scale_pos_weight = neg / pos
print("scale_pos_weight:", scale_pos_weight)
```

```python
xgb = XGBClassifier(random_state=42,use_label_encoder=False,eval_metric='logloss',scale_pos_weight=scale_pos_weight)

grid_search_xgb_weighted = GridSearchCV(xgb,param_grid,cv=5,scoring='f1',n_jobs=-1).fit(X_train, y_train)


best_xgb_weighted = grid_search_xgb_weighted.best_estimator_

y_pred = best_xgb_weighted.predict(X_test)
y_prob = best_xgb_weighted.predict_proba(X_test)[:,1]



print("3. XGBoost with scale_pos_weight and Grid Search")
print("Best parameters:", grid_search_xgb_weighted.best_params_)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1-score:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))

cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)
```

### XGBoost Random Search

```python

xgb = XGBClassifier(random_state=42,eval_metric='logloss',use_label_encoder=False)

param_dist = {
    'n_estimators': [100, 200, 300, 500],
    'max_depth': [3, 5, 6, 8, 10],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'subsample': [0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.8, 0.9, 1.0],
    'min_child_weight': [1, 3, 5, 7],
    'gamma': [0, 0.1, 0.3, 0.5],
    'reg_alpha': [0, 0.1, 1],
    'reg_lambda': [1, 1.5, 2]
}

random_search = RandomizedSearchCV(xgb, param_distributions=param_dist, n_iter=30,  scoring='f1', cv=5,verbose=1,random_state=42,n_jobs=-1).fit(X_train, y_train)

best_xgb_rand = random_search.best_estimator_

print("Best parameters:", random_search.best_params_)
```

```python
y_pred = best_xgb_rand.predict(X_test)
y_prob = best_xgb_rand.predict_proba(X_test)[:,1]

print("XGBoost Random Search")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1:", f1_score(y_test, y_pred))
print("ROC-AUC:", roc_auc_score(y_test, y_prob))
```

```python
import numpy as np
from sklearn.metrics import f1_score

thresholds = np.arange(0.3, 0.7, 0.01)

best_f1 = 0
best_t = 0

for t in thresholds:
    y_pred_t = (y_prob >= t).astype(int)
    f1 = f1_score(y_test, y_pred_t)
    
    if f1 > best_f1:
        best_f1 = f1
        best_t = t

print("Best threshold:", best_t)
print("Best F1:", best_f1)
```

Baseline XGBoost

Базова модель XGBoost продемонструвала високі результати вже без додаткової оптимізації. Значення accuracy становить 0.923, що є порівнянним із Random Forest. Показники precision = 0.817 та recall = 0.836 свідчать про збалансовану здатність моделі як точно визначати події buzz, так і знаходити їх у даних.

F1-score дорівнює 0.827, що є трохи вищим, ніж у базової моделі Random Forest, а ROC-AUC = 0.969 підтверджує високу якість класифікації. Матриця помилок показує відносно невелику кількість як хибно позитивних, так і хибно негативних прогнозів, що свідчить про стабільну роботу моделі.

XGBoost з Grid Search

Оптимізація гіперпараметрів призвела до помітного покращення результатів. Accuracy зросла до 0.932, що є найвищим показником серед усіх XGBoost моделей. Також значно покращилися precision (0.847) та recall (0.845), що призвело до зростання F1-score до 0.846.

Це є найкращим значенням F1-score серед усіх моделей (включаючи Random Forest), що свідчить про оптимальний баланс між точністю та повнотою. Значення ROC-AUC = 0.973 також є найвищим, що підтверджує відмінну здатність моделі розрізняти класи.

Таким чином, XGBoost з оптимізацією гіперпараметрів демонструє найкращу загальну ефективність на даному датасеті.

XGBoost з scale_pos_weight та Grid Search

Використання параметра scale_pos_weight дозволило значно покращити здатність моделі виявляти події buzz. Recall зріс до 0.902, що є найвищим значенням серед усіх моделей, включаючи Random Forest.

Однак це досягнуто за рахунок зниження precision до 0.753, що свідчить про збільшення кількості хибно позитивних прогнозів. Відповідно, accuracy знизилася до 0.913, а F1-score становить 0.821, що є трохи нижчим за оптимізовану модель без врахування ваг.

Значення ROC-AUC = 0.972 залишається високим, що підтверджує стабільність моделі. Матриця помилок показує суттєве зменшення кількості хибно негативних прогнозів (34), що є ключовим результатом при роботі з рідкісними подіями.

```python
import pandas as pd
import matplotlib.pyplot as plt

importances = best_xgb.feature_importances_
feature_names = X_train.columns


feat_imp = pd.DataFrame({
    'feature': feature_names,
    'importance': importances
})

feat_imp['base_feature'] = feat_imp['feature'].apply(lambda x: x.rsplit('_', 1)[0])


grouped_importance = feat_imp.groupby('base_feature')['importance'].sum().reset_index()

grouped_importance = grouped_importance.sort_values(by='importance', ascending=False)

top_grouped = grouped_importance.head(5)

plt.figure()
plt.barh(top_grouped['base_feature'], top_grouped['importance'])
plt.gca().invert_yaxis()
plt.title("Top Feature Groups (Aggregated over Time)")
plt.xlabel("Total Importance")
plt.show()
```

```python
import numpy as np
from sklearn.metrics import f1_score

thresholds = np.arange(0.3, 0.7, 0.01)

best_f1 = 0
best_t = 0

for t in thresholds:
    y_pred_t = (y_prob >= t).astype(int)
    f1 = f1_score(y_test, y_pred_t)
    
    if f1 > best_f1:
        best_f1 = f1
        best_t = t

print("Best threshold:", best_t)
print("Best F1:", best_f1)
```

Baseline XGBoost

Базова модель XGBoost продемонструвала дуже високі результати. Значення accuracy становить 0.968, що є найвищим серед усіх моделей для цього датасету. Також спостерігається високе значення precision = 0.877, що означає, що більшість передбачених подій buzz є правильними.

Recall дорівнює 0.810, що свідчить про те, що частина реальних подій buzz все ще пропускається. Водночас F1-score становить 0.842, що є дуже хорошим результатом. Значення ROC-AUC = 0.989 підтверджує відмінну здатність моделі розрізняти класи.

Матриця помилок показує невелику кількість помилок: лише 19 хибно позитивних і 32 хибно негативних прогнозів, що свідчить про стабільну та ефективну роботу моделі.

XGBoost з Grid Search

Оптимізація гіперпараметрів не призвела до покращення результатів. Accuracy становить 0.965, що трохи нижче за baseline. Precision = 0.855 також знизився, тоді як recall залишився на рівні 0.810.

F1-score дорівнює 0.832, що є нижчим за базову модель. Водночас ROC-AUC = 0.989 є трохи вищим, що вказує на хорошу здатність до розділення класів, але без покращення практичної ефективності.

Таким чином, Grid Search у цьому випадку не дав суттєвого виграшу.

XGBoost з scale_pos_weight та Grid Search

Використання параметра scale_pos_weight дозволило досягти найкращих результатів для цього датасету. Recall зріс до 0.869, що є найвищим серед усіх моделей, що означає значне покращення виявлення подій buzz.

При цьому precision становить 0.839, що є трохи нижчим за baseline, але все ще на високому рівні. F1-score дорівнює 0.854, що є найкращим результатом серед усіх варіантів XGBoost для цього датасету.

Accuracy становить 0.968, що навіть трохи перевищує базову модель, а ROC-AUC = 0.990 є найвищим показником серед усіх експериментів.

Аналіз матриці помилок показує зменшення кількості хибно негативних прогнозів до 22, що є ключовим результатом для задачі виявлення рідкісних подій.

```python
importances = best_xgb.feature_importances_

feature_names = X_train.columns

feat_imp = pd.DataFrame({
    'feature': feature_names,
    'importance': importances
})

feat_imp['base_feature'] = feat_imp['feature'].apply(lambda x: x.rsplit('_', 1)[0])

grouped_importance = feat_imp.groupby('base_feature')['importance'].sum().reset_index()

grouped_importance = grouped_importance.sort_values(by='importance', ascending=False)

top_grouped = grouped_importance.head(5)

plt.figure()
plt.barh(top_grouped['base_feature'], top_grouped['importance'])
plt.gca().invert_yaxis()
plt.title("Top Feature Groups (Aggregated over Time)")
plt.xlabel("Total Importance")
plt.show()
```

### Result 

Отримані результати показали, що XGBoost демонструє високу та стабільну якість класифікації на всіх трьох датасетах. Навіть базова модель без додаткової оптимізації забезпечує високі значення accuracy, F1-score та ROC-AUC, що свідчить про здатність моделі ефективно виявляти складні нелінійні залежності в даних.

Для датасету TH_500 найкращі результати продемонструвала модель XGBoost з Grid Search, яка досягла найвищого значення F1-score та ROC-AUC, що свідчить про оптимальний баланс між precision та recall. Використання параметра scale_pos_weight дозволило суттєво підвищити recall, однак за рахунок зниження precision.

Для датасету TH_1000 базова модель XGBoost вже показує дуже високі результати, і додаткова оптимізація гіперпараметрів не призводить до значного покращення. У цьому випадку найкращий баланс між метриками досягається при використанні scale_pos_weight, що дозволяє покращити виявлення подій buzz.

Для найбільш незбалансованого датасету TH_1500 найефективнішою виявилася модель XGBoost з урахуванням ваг класів, яка забезпечила найвищі значення F1-score та ROC-AUC. Це підтверджує, що при значному дисбалансі класів особливо важливо враховувати вагу позитивного класу.

Загалом можна зробити висновок, що оптимізація гіперпараметрів є найбільш ефективною для менш незбалансованих даних (σ = 500), тоді як при зростанні незбалансованості (σ = 1000, 1500) ключову роль відіграє врахування ваг класів за допомогою параметра scale_pos_weight.

Таким чином, модель XGBoost є ефективним інструментом для задачі прогнозування buzz, а її найкраща конфігурація залежить від рівня незбалансованості даних: при помірному дисбалансі доцільно використовувати оптимізацію параметрів, тоді як при сильному дисбалансі — методи врахування ваг класів.
